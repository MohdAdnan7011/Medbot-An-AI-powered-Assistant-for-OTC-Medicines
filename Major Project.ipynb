```python
import numpy as np
import pandas as pd

data = pd.read_csv("Over_the_counter.csv")
```

```python
data.head(25)
```

```python

data['Dosage_Form_Count'] = data.groupby('Dosage Form')['Dosage Form'].transform('count')
```

```python
data['Dosage_Form_Count']
```

```python
# Extract the numeric part of the 'Strength' column
data['Strength_Numeric'] = data['Strength'].str.extract('(\d+)').astype(float)
```

```python
data['Strength_Numeric']
```

```python
# Create bins for strength
bins = [0, 200, 500, 1000, 2000] # Define your own bins
labels = ['Very Low', 'Low', 'Medium', 'High']
data['Strength_Binned'] = pd.cut(data['Strength_Numeric'], bins=bins, labels=labels, right=False)
```

```python
data['Strength_Binned']
```

```python
# Count the number of indications for each drug name
data['Indications_Count'] = data.groupby('Name')['Indication'].transform('count')
```

```python
data['Indications_Count']
```

```python
# Combine two columns to create a new one
data['Category_Dosage_Interaction'] = data['Category'] + '_' + data['Dosage Form']
```

```python
data['Category_Dosage_Interaction']
```

```python
# Create a binary feature for liquid dosage forms
liquid_forms = ['Syrup', 'Injection', 'Liquid', 'Solution'] # Add any other liquid forms
data['Is_Liquid'] = data['Dosage Form'].apply(lambda x: 1 if x in liquid_forms else 0)
```

```python
data['Is_Liquid']
```

```python
data.head()
```

```python
# Count the number of drugs per manufacturer
data['Manufacturer_Drug_Count'] = data.groupby('Manufacturer')['Name'].transform('count')
```

```python
data['Manufacturer_Drug_Count']
```

```python
# Calculate the length of the 'Name' string
data['Name_Length'] = data['Name'].apply(len)
```

```python
data['Name_Length']
```

```python
print("Step 1: Loading mock dataset based on your provided images...")
mock_data = {
    'Name': ['Acetocillin', 'Ibuprocillin', 'Amoxicillin', 'Metovir', 'Ibuprovir', 'Dextrovir', 'Dextronazole', 'Dextrovir', 'Dextrostatin', 'Dolomet'] * 2500,
    'Category': ['Antidiabetic', 'Antiviral', 'Antifungal', 'Antipyretic', 'Antidepressant', 'Antipyretic', 'Analgesic', 'Antipyretic', 'Antibiotic', 'Antipyretic'] * 2500,
    'Dosage Form': ['Cream', 'Injection', 'Tablet', 'Ointment', 'Syrup', 'Syrup', 'Ointment', 'Syrup', 'Inhaler', 'Ointment'] * 2500,
    'Strength': ['938 mg', '337 mg', '802 mg', '641 mg', '758 mg', '920 mg', '405 mg', '327 mg', '885 mg', '426 mg'] * 2500,
    'Manufacturer': ['Roche Holding AG', 'CSL Limited', 'Teva Pharmaceutical Industries Ltd.', 'Takeda Pharmaceutical Company Limited', 'Eli Lilly and Company', 'Novo Nordisk A/S', 'Eli Lilly and Company', 'Gilead Sciences, Inc.', 'AbbVie Inc.', 'Teva Pharmaceutical Industries Ltd.'] * 2500,
    'Indication': ['Virus', 'Infection', 'Wound', 'Infection', 'Fungus', 'Fever', 'Pain', 'Fever', 'Wound', 'Fever'] * 2500,
    'Classification': ['Over-the-Counter'] * 25000,
}
```

```python
data = pd.DataFrame(mock_data)
print("Initial dataset loaded.")
#print(data.head(10))
data.head(20)
```

```python
# Step 2: Drop irrelevant or redundant columns
data.drop(columns=['Classification'], inplace=True)
```

```python
data.head(5)
```

```python
# Step 3: Standardize text data and handle case sensitivity
print("\nStep 3: Standardizing text data...")
for col in ['Category', 'Dosage Form', 'Manufacturer', 'Indication']:
    data[col] = data[col].str.strip().str.title()
print("Text standardization complete.")
data.head()
```

```python
# Step 4: Clean 'Strength' column and extract numeric value
print("\nStep 4: Cleaning and converting 'Strength' to numeric...")
data['Strength_Numeric'] = data['Strength'].str.extract('(\d+)').astype(float)
data['Strength_Unit'] = data['Strength'].str.extract('([a-zA-Z\/]+)')
print("Numeric strength and unit extracted.")
data[['Strength', 'Strength_Numeric', 'Strength_Unit']].head()
```

```python
safety_concerns = {
    'Injection': 3, 'Inhaler': 2, 'Ointment': 1, 'Cream': 1, 'Tablet': 1, 'Syrup': 1,
}
data['Safety_Score'] = data['Dosage Form'].map(safety_concerns).fillna(0)
```

```python
data.head(20)
```

```python
print("\nStep 18: Visualizing Safety Score by Dosage Form...")
plt.figure(figsize=(12, 8))
sns.boxplot(x='Dosage Form', y='Safety_Score', data=data, palette='tab10')
plt.title('Safety Score Distribution by Dosage Form')
plt.xlabel('Dosage Form')
plt.ylabel('Safety Score')
plt.tight_layout()
plt.show()
```

```python
print("Counting the occurrences of each Safety Score...")
safety_score_counts = data['Safety_Score'].value_counts().sort_index()

# Display the results
print(safety_score_counts)
```

```python
print("\nGenerating bar plot for Safety Score counts...")
plt.figure(figsize=(10, 6))
sns.barplot(x=safety_score_counts.index, y=safety_score_counts.values, palette='viridis')


plt.title('Count of Medications by Safety Score')
plt.xlabel('Safety Score (1 = Low Risk, 2 = Medium Risk, 3 = High Risk)')
plt.ylabel('Number of Medications')

# Display the plot
plt.tight_layout()
plt.show()
```
